```python
%sql
USE SCHEMA framework_testes; --definir qual é o schema

DROP TABLE IF EXISTS test_out_results;
CREATE TABLE IF NOT EXISTS test_out_results (
  RESULT_ID INT,
  TEST_ID INT,
  QUERY STRING,
  RESULT STRING,
  EXECUTION_PARAMETER STRING,
  TIMESTAMP TIMESTAMP
);

DROP TABLE IF EXISTS test_out_results_details;
CREATE TABLE IF NOT EXISTS test_out_results_details (
  DETAILS_ID INT,
  RESULT_ID INT,
  KEY_FIELDS STRING,
  TIMESTAMP TIMESTAMP
);

DROP TABLE IF EXISTS test_config_parameters;
CREATE TABLE IF NOT EXISTS test_config_parameters (
  TEST_ID INT,
  TYPE_ID STRING,
  SUBTYPE_ID STRING,
  SOURCE_TABLE STRING,
  DEST_TABLE STRING,
  JOIN_KEY STRING,
  SOURCE_SELECT_FIELD STRING,
  DEST_SELECT_FIELD STRING,
  SOURCE_FILTER STRING,
  DEST_FILTER STRING,
  SOURCE_GROUPBY STRING,
  DEST_GROUPBY STRING,
  KEY_FIELDS STRING
);

DROP TABLE IF EXISTS test_rel_set_parameter;
CREATE TABLE IF NOT EXISTS test_rel_set_parameter (
  SET_ID INT,
  TEST_ID INT
);

DROP TABLE IF EXISTS test_config_set;
CREATE TABLE IF NOT EXISTS test_config_set (
  SET_ID INT,
  NAME STRING,
  DESCRIPTION STRING
);

DROP TABLE IF EXISTS test_config_subtype;
CREATE TABLE IF NOT EXISTS test_config_subtype (
  SUBTYPE_ID INT,
  NAME STRING,
  DESCRIPTION STRING
);

DROP TABLE IF EXISTS test_config_type;
CREATE TABLE IF NOT EXISTS test_config_type (
  TYPE_ID INT,
  NAME STRING,
  DESCRIPTION STRING
);

DROP TABLE IF EXISTS test_rel_type_subtype;
CREATE TABLE IF NOT EXISTS test_rel_type_subtype (
  TYPE_ID INT,
  SUBTYPE_ID INT,
  NAME STRING,
  DESCRIPTION STRING
);
```
